```python
import pandas as pd
```

```python
data=pd.read_csv("Salary_dataset.csv")
data.head()
```

```python
data.drop("Unnamed: 0", axis=1, inplace=True)
data.head()
```

```python
x=data.iloc[:,:-1]
x.head()
```

```python
y = data["Salary"]
y
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test= train_test_split(x,y, test_size=0.2)
x_train.shape
```

```python
y_train.shape
```

```python
x_test.shape
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
model.fit(x_train, y_train)
LinearRegression()
```

```python
from sklearn.metrics import r2_score
```

```python
model.predict([[1.2]])
```

```python
y_pred=model.predict(x_test)
```

```python
r2_score(y_test, y_pred) *100
```

```python
plt.scatter(x_train, y_train, color="green", label="Training Data")
plt.scatter(x_test, y_test, color="blue", label="Training Data")
plt.plot(x_test, model.predict(x_test), color="red", label="Best fit Line")
plt.legend()
plt.show()
```

```python
plt.scatter(x_train, y_train, color="green", label="Training Data")
plt.scatter(x_test, y_test, color="blue", label="Training Data")
plt.plot(x_train, model.predict(x_train), color="red", label="Best fit Line")
plt.legend()
plt.show()
```

```python
evaluation=pd.DataFrame({"Actual Value" : y_test, "predicted Value" : y_pred, "bias/Error" : y_test - y_pred})
evaluation
```
